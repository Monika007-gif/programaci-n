```python
precios_frutas = {
                "manzana": 2.35,
                "pera": 3.05,
                "naranja": 1.75,
                "limon": 1.95,
                "melon": 4.85,
                "sandia": 3.45,
                "platano": 2.95,
                "cereza": 5.90,
                "uva": 4.15,
                "fresa": 4.25,
                }

print(precios_frutas)

precios_frutas["fresa"] = 6.54

print(precios_frutas)

for fruta in precios_frutas:
    precios_frutas[fruta] = round(precios_frutas[fruta] * 1.10, 2) # incrementamos el precio de la fruta en un 10%

print(precios_frutas)
```

# Funciones, len devuelve una lista con las claves del diccionario.
# Con in y not in sabemos si una clave está o no en el diccionario

```python
telefonos = {
            "Juan": 123456789,
            "Pedro": 987654321,
            "Ana": 654987321,
            "Maria": 321654987,
            }

print(len(telefonos))
print("Juan" in telefonos)
print("Andrea" in telefonos)
```

# clear elimina todos los elementos del diccionario, lo deja vacío

```python
precios_frutas = {
                "manzana": 2.35,
                "pera": 3.05,
                "naranja": 1.75,
                "limon": 1.95,
                "melon": 4.85,
                "sandia": 3.45,
                "platano": 2.95,
                "cereza": 5.90,
                "uva": 4.15,
                "fresa": 4.25,
                }

print(precios_frutas)

precios_frutas.clear() # Elimina todos los elementos del diccionario

# print(precios_frutas)
```

# pop(clave): elimina el elemento con la clave indicada y devuelve su valor.

```python
precios_frutas = {
                "manzana": 2.35,
                "pera": 3.05,
                "naranja": 1.75,
                "limon": 1.95,
                "melon": 4.85,
                "sandia": 3.45,
                "platano": 2.95,
                "cereza": 5.90,
                "uva": 4.15,
                "fresa": 4.25,
                }

precio = precios_frutas.pop("manzana") # Elimina el elemento con clave "manzana" y devuelve su valor

print(precio)
print(precios_frutas)
```

# popitem() elimina el último elemento insertado y devuelve su clase y su valor

```python
print(precios_frutas.popitem()) 
```

# update(diccionario): añade los pares clave-valor del diccionario indicado y modifica los existentes.

```python
diccionario = {
            "hola":"привіт", 
            "adios":"бувай",
            "casa":"дім",
            "perro":"пес",
            "gato":"кіт",
            }

mas_palabras = {
            "coche":"автомобіль",
            "libro":"книга",
            "ordenador":"комп'ютер",
            "moto":"мотоцикл",
            "casa":"будинок",
            }

print(diccionario)

diccionario.update(mas_palabras) # Añade al diccionario las claves y valores del diccionario mas_palabras

print(diccionario)
```
