# Con una lista podemos contar cuántos elementos tiene, unir dos listas y repetir los elementos de una lista, len es una funcion que puede usarse con listas, cadenas y otros tipos de datos. Con una lista sencilla, en este caso devuelve 3 porque la lista contiene tres elementos, con una lista con otra lista dentro, devuelve 3 y no 4 porque len() cuenta los elementos del nivel exterior de la lista. El primero es 'one', el segundo [2,3] y el tercero 'four', la lista de dentro con dos elementos, se cuenta como un único elemento de la lista exterior. Len cuenta los elementos de la lista, no todos los elementos dentro de listas anidadas. El operador + permite unir dos listas. El operador * hace que se repita, en este caso 4 veces
